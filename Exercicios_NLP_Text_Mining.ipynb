```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Exercício 1: Download do dataset
url = 'https://raw.githubusercontent.com/watinha/nlp-text-mining-datasets/main/b2w-50.csv'
df = pd.read_csv(url)
print(f'Dataset carregado com {len(df)} registros')
print(f'Colunas: {list(df.columns)}')
df.head()
```

## Exercício 2 - CountVectorizer

Carregar o dataset, extrair o corpus a partir do atributo `review_text` e contar as seguintes propriedades do texto usando a API CountVectorizer:
- Palavra que mais aparece e a que menos aparece no texto
- Média de palavras por documento

```python
# Exercício 2: Extrair corpus e usar CountVectorizer

# Extrair o corpus do atributo review_text
corpus = df['review_text'].values.astype(str)
print(f'Corpus com {len(corpus)} documentos\n')

# Aplicar CountVectorizer
count_vectorizer = CountVectorizer()
count_matrix = count_vectorizer.fit_transform(corpus)

# Obter nomes dos termos
feature_names = count_vectorizer.get_feature_names_out()
print(f'Total de termos no vocabulário: {len(feature_names)}')

# Somar frequências de cada termo em todo o corpus
term_frequencies = np.asarray(count_matrix.sum(axis=0)).flatten()

# Criar DataFrame com termos e frequências
freq_df = pd.DataFrame({'termo': feature_names, 'frequencia': term_frequencies})
freq_df = freq_df.sort_values('frequencia', ascending=False)

# Palavra mais frequente
most_frequent = freq_df.iloc[0]
print(f'\nPalavra MAIS frequente: "{most_frequent["termo"]}" ({most_frequent["frequencia"]} ocorrências)')

# Palavra menos frequente
least_frequent = freq_df.iloc[-1]
print(f'Palavra MENOS frequente: "{least_frequent["termo"]}" ({least_frequent["frequencia"]} ocorrência(s))')

# Média de palavras por documento
words_per_doc = np.asarray(count_matrix.sum(axis=1)).flatten()
avg_words = words_per_doc.mean()
print(f'\nMédia de palavras por documento: {avg_words:.2f}')

print('\n--- Top 10 termos mais frequentes ---')
print(freq_df.head(10).to_string(index=False))

print('\n--- 10 termos menos frequentes ---')
print(freq_df.tail(10).to_string(index=False))
```

## Exercício 3 - Cálculo do IDF

Calcular o IDF do corpus e identificar os 5 termos com maior e menor valor de IDF.

```python
# Exercício 3: Calcular IDF usando TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)

# Obter nomes dos termos e valores de IDF
tfidf_features = tfidf_vectorizer.get_feature_names_out()
idf_values = tfidf_vectorizer.idf_

# Criar DataFrame com termos e IDF
idf_df = pd.DataFrame({'termo': tfidf_features, 'idf': idf_values})
idf_df = idf_df.sort_values('idf', ascending=False)

print('=== 5 termos com MAIOR IDF (mais raros) ===')
print(idf_df.head(5).to_string(index=False))

print('\n=== 5 termos com MENOR IDF (mais comuns) ===')
print(idf_df.tail(5).to_string(index=False))
```
